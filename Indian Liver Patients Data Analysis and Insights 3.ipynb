# Indian Liver Patients Data Analysis and Insights

## Introduction to Indian Liver Patient Analysis

**Liver disease** is a major global health concern that affects the liver's ability to perform vital functions such as filtering toxins, producing bile, and regulating metabolism. Early detection and diagnosis of liver disorders are crucial, as delayed treatment can lead to severe complications such as cirrhosis, liver failure, or cancer.

The purpose of this analysis is to explore and understand the key clinical factors associated with liver disease and present actionable insights through clear and meaningful data visualization. This project uses Python libraries such as **Pandas**, **NumPy**, **Matplotlib**, and **Seaborn** to clean, analyze, and visualize the Indian Liver Patient Dataset (ILPD).

## Data Source

The dataset used in this project is the **Indian Liver Patient Dataset (ILPD)**, originally collected from the northeast of Andhra Pradesh, India, and hosted on the UCI Machine Learning Repository / Kaggle. It contains clinical records of **583 patients** (416 liver patient records and 167 non-liver patient records).

**Note:** Two additional calculated clinical features—**Indirect Bilirubin** and **SGOT/SGPT Ratio**—were derived from existing attributes to enhance the medical depth of the analysis.

## Attribute Details

| Attribute Name | Description |
| :--- | :--- |
| **Age** | Age of the patient (in years). |
| **Gender** | Gender of the patient (Male / Female). |
| **Total_Bilirubin** | Total amount of bilirubin in the blood (mg/dL). High levels indicate liver dysfunction. |
| **Direct_Bilirubin** | Conjugated bilirubin level in the blood (mg/dL). |
| **Indirect_Bilirubin** | Unconjugated bilirubin calculated as `Total Bilirubin - Direct Bilirubin` (mg/dL). |
| **Alkaline_Phosphatase** | Enzyme level in the blood (IU/L) associated with bile ducts and liver health. |
| **SGPT_Alamine_Aminotransferase** | Liver enzyme (IU/L) that rises significantly during liver cell damage. |
| **SGOT_Aspartate_Aminotransferase** | Enzyme (IU/L) found in the liver and other tissues; elevated in liver injury. |
| **Total_Proteins** | Total protein concentration in the blood (g/dL), including albumin and globulin. |
| **Albumin** | Essential protein produced by the liver (g/dL); low levels suggest chronic liver disease. |
| **Albumin_Globulin_Ratio** | Ratio of albumin to globulin in the blood; helps assess liver and kidney health. |
| **SGOT_SGPT_Ratio** | Ratio of SGOT to SGPT enzymes, useful for differentiating types of liver damage. |
| **Liver_Disease_Status** | Target variable indicating diagnosis (`1` = Liver Patient, `2` = Non-Liver Patient). |

## Data Cleaning and Preprocessing

This phase addresses data quality issues including duplicate records, missing values, incorrect data types, formatting inconsistencies, outliers, and logical discrepancies.

### Step 1: Remove Duplicate Records

Checking for and removing duplicate patient records to ensure data integrity and prevent skewed analysis.

```python
# Check the number of duplicate records
print("Duplicate rows before removal:", df.duplicated().sum())

# Remove duplicate records
df.drop_duplicates(inplace=True)

# Verify removal and check the updated dataset shape
print("Duplicate rows after removal:", df.duplicated().sum())
print("Updated dataset shape:", df.shape)
```

### Step 2: Handle Missing Values (NULLs)

Identifying missing values across all columns and imputing the missing entries in `A/G Ratio` using the median to preserve the data distribution.

```python
# Check for missing values in each column
print("Missing values before handling:\n", df.isnull().sum())

# Fill missing values in 'A/G Ratio' with the median
df['A/G Ratio'] = df['A/G Ratio'].fillna(df['A/G Ratio'].median())

# Verify that all missing values are handled
print("\nMissing values after handling:\n", df.isnull().sum())
```

### Step 3: Clean Column Spaces and Correct Data Types

Stripping leading/trailing spaces from column headers and converting columns (`Age`, `is_Patient`, `Gender`) into their appropriate data types for efficient processing.

```python
# Remove any leading or trailing spaces from column names
df.columns = df.columns.str.strip()

# Check current data types of all columns
print("Data types before correction:\n", df.dtypes)

# Ensure appropriate data types
df['Age'] = df['Age'].astype(int)
df['is_Patient'] = df['is_Patient'].astype(int)
df['Gender'] = df['Gender'].astype('category')

# Verify updated data types
print("\nData types after correction:\n", df.dtypes)
```

### Step 4: Check and Remove Irrelevant Columns

Verifying all dataset features to ensure no unnamed or irrelevant columns are present in the DataFrame.

```python
# Check all existing columns
print("Current columns:", df.columns.tolist())

# Drop any unnamed or irrelevant columns if present
irrelevant_cols = [col for col in df.columns if 'Unnamed' in col]
df.drop(columns=irrelevant_cols, inplace=True)

# Verify remaining columns
print("Irrelevant columns removed:", irrelevant_cols)
print("Remaining columns count:", df.shape[1])
```

### Step 5: Standardize Text Formats and Correct Spelling Errors

Standardizing categorical text values in the `Gender` column and correcting the spelling error in the `Alkaline Phosphatase` column header.

```python
# Check unique values in 'Gender' before standardization
print("Unique values in Gender before cleaning:", df['Gender'].unique().tolist())

# Remove unnecessary spaces, standardize text format, and fix any spelling inconsistencies
df['Gender'] = df['Gender'].astype(str).str.strip().str.title()
df['Gender'] = df['Gender'].replace({'M': 'Male', 'F': 'Female', 'Femal': 'Female', 'Mal': 'Male'})
df['Gender'] = df['Gender'].astype('category')

# Correct spelling error in column name ('Phosphotase' -> 'Phosphatase')
df.rename(columns={'Alkaline Phosphotase': 'Alkaline Phosphatase'}, inplace=True)

# Verify cleaned Gender values and updated column names
print("Unique values in Gender after cleaning:", df['Gender'].unique().tolist())
print("Updated column names:", df.columns.tolist())
```

### Step 6: Detect and Handle Outliers

Applying the **Interquartile Range (IQR) Capping** method to handle extreme outliers in skewed clinical columns (`Total Bilirubin`, `Direct Bilirubin`, `Alkaline Phosphatase`, `Sgpt`, and `Sgot`) without losing valuable patient records.

```python
# Select skewed numerical columns to check and handle outliers
outlier_cols = ['Total Bilirubin', 'Direct Bilirubin', 'Alkaline Phosphatase', 'Sgpt', 'Sgot']

# Detect and cap outliers using the IQR method
for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    outliers_count = ((df[col] < lower_limit) | (df[col] > upper_limit)).sum()
    print(f"{col}: {outliers_count} outliers detected and capped")
    
    # Cap values to the upper and lower limits
    df[col] = df[col].clip(lower=lower_limit, upper=upper_limit)

# Verify summary statistics after outlier handling
df[outlier_cols].describe().round(2)
```

### Step 7: Validate Data Accuracy and Recalculate Derived Features

Validating logical constraints (age range, target labels, negative values, and bilirubin consistency) and recalculating derived columns (`Indirect Bilirubin` and `SGOT/SGPT Ratio`) after outlier capping.

```python
# Validate Age range (must be between 1 and 120)
invalid_age = df[(df['Age'] <= 0) | (df['Age'] > 120)]
print("Invalid Age records:", len(invalid_age))

# Validate target column 'is_Patient' (should only contain 1 and 2)
print("Unique values in is_Patient:", df['is_Patient'].unique().tolist())

# Ensure Direct Bilirubin does not exceed Total Bilirubin
invalid_bilirubin = df[df['Direct Bilirubin'] > df['Total Bilirubin']]
print("Records where Direct Bilirubin > Total Bilirubin:", len(invalid_bilirubin))

# Recalculate derived columns to maintain accuracy after outlier capping
df['Indirect Bilirubin'] = (df['Total Bilirubin'] - df['Direct Bilirubin']).clip(lower=0).round(2)
df['SGOT/SGPT Ratio'] = (df['Sgot'] / df['Sgpt']).round(2)

# Verify no negative values exist in numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
print("Negative values count per column:\n", (df[num_cols] < 0).sum())
```

### Step 8: Resolve Discrepancies, Rename Columns, and Export Cleaned Data

Correcting minor records where Direct Bilirubin exceeded Total Bilirubin, renaming columns for analytical clarity, and saving the final cleaned dataset as `cleaned_liver_patient_dataset.csv`.

```python
# Fix the 3 records where Direct Bilirubin exceeds Total Bilirubin
df['Direct Bilirubin'] = df[['Direct Bilirubin', 'Total Bilirubin']].min(axis=1)
df['Indirect Bilirubin'] = (df['Total Bilirubin'] - df['Direct Bilirubin']).round(2)

# Rename columns for clarity and consistency
df.rename(columns={
    'Total Bilirubin': 'Total_Bilirubin',
    'Direct Bilirubin': 'Direct_Bilirubin',
    'Indirect Bilirubin': 'Indirect_Bilirubin',
    'Alkaline Phosphatase': 'Alkaline_Phosphatase',
    'Sgpt': 'SGPT_Alamine_Aminotransferase',
    'Sgot': 'SGOT_Aspartate_Aminotransferase',
    'Total Proteins': 'Total_Proteins',
    'A/G Ratio': 'Albumin_Globulin_Ratio',
    'SGOT/SGPT Ratio': 'SGOT_SGPT_Ratio',
    'is_Patient': 'Liver_Disease_Status'
}, inplace=True)

# Save the cleaned dataset to a new CSV file
df.to_csv("cleaned_liver_patient_dataset.csv", index=False)

# Display the final cleaned dataset overview
print("Data cleaning completed successfully! Final shape:", df.shape)
df.head()
```

✔ **Data Preprocessing was successfully completed!!** Ensuring the dataset is clean, consistent, and ready for analysis.
